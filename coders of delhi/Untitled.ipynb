```python
import json
```

```python
def loaddata(filename):
    with open(filename,"r") as file:
        data=json.load(file)
    return data
```

```python
data=loaddata("./data.json")
data
```

```python
type(data)
```

```python
def display_details(data):
    print("USER DETAILS : ")
    for user in data["users"]:
        print(f"ID : {user['id']} , NAME : {user['name']} , FRIENDS : {user['friends']} , LIKED PAGES : {user['liked_pages']}")
    print("\PAGES DETAILS: ")
    for page in data["pages"]:
        print(f"PAGE ID : {page['id']} , PAGE NAME : {page['name']}")
display_details(data)
```
