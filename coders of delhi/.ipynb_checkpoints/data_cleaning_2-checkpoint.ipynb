```python
import json

def clean_data(data):
    #remoing no name data
    data["users"]=[user for user in data["users"] if user["name"].strip()]
    print(data)
    return data



def load_data(file):
    with open(file,"r") as file:
        data=json.load(file)
        return data
data=load_data("data2.json")
data=clean_data(data)
json.dump(data,open("cleaned_data.json","w"),indent=4)

```
