```python
class student:
    name='pranay'
    def __init__(self,name,section):
        self.name=name
        self.section=section
    def details(self):
        print(f"name of student : {self.name} \nsection : {self.section}")
    @classmethod
    def classmethod(cls):
        print(f"name without self : {cls.name}")
c=student("pranay","B")
c.details()
c.classmethod()
```
