```python
square=lambda x:x*x
square(5)
```

```python
cube=lambda x:x*x*x
cube(5)
```

```python
add=lambda a,b:a+b
add(1,5)
```

```python
def calc(x):
    a=int(input("1 for square,2 for cube"))
    match(a):
        case 1:
            print(f"{square(x)}")
        case 2:
            print(f"{cube(x)}")
        case _:
            print("invalid choice")
calc(5)
```
